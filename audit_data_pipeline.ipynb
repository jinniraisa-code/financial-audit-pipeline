```python
import numpy as np
import pandas as pd

# 1. Load raw enterprise transaction batch into a structured DataFrame
data = {
    "Invoice_ID": ["INV101", "INV102", "INV103", None, "INV105", "INV101"],
    "Vendor": [
        "Vendor A",
        "Vendor B",
        "Vendor C",
        "Vendor D",
        "Vendor A",
        "Vendor A",
    ],
    "Amount": [12500.00, 450.00, -100.00, 8900.00, 150000.00, 12500.00],
    "Posting_Date": [
        "2026-09-01",
        "2026-09-03",
        "2026-09-05",
        "2026-09-05",
        "2026-09-10",
        "2026-09-01",
    ],
}
df = pd.DataFrame(data)

# 2. Automated Internal Control & Audit Risk Checks

# Check A: Flag duplicate invoice numbers (Risk: Double payments/duplicate billing)
duplicates = df[df.duplicated(subset=["Invoice_ID"], keep=False)]

# Check B: Identify missing transaction reference IDs (Risk: Unmapped subledger / missing compliance)
missing_ids = df[df["Invoice_ID"].isna()]

# Check C: Isolate negative transaction values (Risk: Unauthorized credit memos / posting errors)
negative_amounts = df[df["Amount"] < 0]

# 3. Output Control Exception Summary
print("Total Transactions Flagged for Review:")
print(f"- Duplicate Invoices: {len(duplicates)}")
print(f"- Missing Invoice IDs: {len(missing_ids)}")
print(f"- Negative Amounts (Reversals/Errors): {len(negative_amounts)}")
```
